# 01 – IBKR Connection Test
Tests connectivity to TWS / IB Gateway.  
**Requirements**: TWS or IB Gateway running on your machine with API access enabled.  
- TWS paper port: **7497**  (live: 7496)
- IB Gateway paper port: **4002** (live: 4001)

Ireland users: your account is held at IBKR Ireland (IBKRIE). Base currency EUR.

```python
import sys
sys.path.insert(0, '..')

from config.settings import IBKR_TWS_PORT, IBKR_HOST, IBKR_CLIENT_ID
print(f'Connecting to {IBKR_HOST}:{IBKR_TWS_PORT} (client_id={IBKR_CLIENT_ID})')
```

```python
from core.ibkr_connection import IBKRConnection

conn = IBKRConnection(port=IBKR_TWS_PORT)
connected = conn.connect(timeout=10)
print('Connected:', connected)
```

```python
# Fetch historical daily data for a European stock
if connected:
    df = conn.get_historical_data(
        symbol='SHEL',
        exchange='LSE',
        currency='GBP',
        duration='6 M',
        bar_size='1 day'
    )
    print(df.tail())
else:
    print('Not connected – skipping data fetch')
```

```python
if connected:
    conn.disconnect()
```
